```python
arr = [3, 7, 1, 9, 0, 5]

def boonli(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = arr[:mid]
    right = arr[mid:]

    leftSort = boonli(left)
    rightSort = boonli(right)
    
    return merge(leftSort, rightSort)

def merge(left, right):
    i, j = 0, 0
    sortedArr = []

    while i < len(left) and j < len(right):
        if left[i] > right[j]:
            sortedArr.append(right[j])
            j += 1
        else:
            sortedArr.append(left[i])
            i += 1
    while j < len(right):
        sortedArr.append(right[j])
        j += 1
    while i < len(left):
        sortedArr.append(left[i])
        i += 1
    return sortedArr

print(boonli(arr))
```
