```python
def recur(arr):
    if len(arr) <= 1:
        return arr
    else:
        return [arr[-1]] + recur(arr[:-1])

arr = input()
arr = list(arr)
print(''.join(recur(arr)))
```

```python
max = 0
def recur(n):
    if len(n) <= 1:
        return n[0]
    else:
        max = recur(n[:-1])
        if max < n[-1]:
            max = n[-1]
        return max
n = list(map(int, input().split()))
print(recur(n))
```

```python
b=[]
c=[]

def recur(n):
    if n == 1:
        c.append(n)
    else:
        b.append(n)
        recur(n-1)
        c.append(n)

n = int(input())
recur(n)
print(b, c)
```

```python
memo = []
def fib(n):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    
    memo[n] = fib(n-1) + fib(n-2)
    return memo[n]
n = int(input())
fib(n)
```

```python
def fib(n):
    dp = [0] * (n + 1)
    dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    return dp[n]
fib(int(input()))
```
